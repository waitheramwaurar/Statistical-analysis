#### Use an appropriate statistical method to test whether the average number of required 
indusrial  training sessions is equal to 2.

```python
# Null hypothesis: mean = 2
# Alternative hypothesis: mean is not 2

# Using the one-sample t-test we will compare the actual mean with the mean given above.
from scipy.stats import ttest_1samp

talent_df = talent_df.rename(columns={'How many separate industrial trainings are you expected to complete in your department?': 'Trainings'})
trainings = talent_df['Trainings'].dropna()

hyp_value = 2
t_statistic, p_value = ttest_1samp(trainings, hyp_value)
p_value
# print(f"T-statistic: {t_statistic:.2f}")
# print(f"P-value: {p_value:.3f}")
```

A one-sample t-test was performed to evaluate whether the average number of require industrial trainings is 2.
The The test returned a p-value of 0.00000375 which is extremely below 0.05 significance level.
This provides strong evidence to reject the null hypothsis, suggesting that the average number of required trainings is significantly different from 2.

#### Apply a suitable inferential statistics method to draw conclusions about student ssatisfactio  with the quality of lectures received.

Where 1 = Very dissatisfied, 2 = Dissatisfied, 3 = Indifferent, 4 = Satisfied, 5 = Very
satisfied

I would like to test whether avergae satisfaction differs from a neutral point which is indeffernce
i.e 3 = Indifferent

Using the one-sample t-test
Null hypothesis: The average satisfaction score is 3 (indifference)
Alternative hypothesis: The average satisfaction is not 3 (so students are either satisfied or dissatisfied)



```python
talent_df = talent_df.rename(columns={'Please rate your satisfaction with the quality of the lectures you receive': 'Satisfaction'})
# talent_df.head()

hyp_value1 = 3
t_stat, p_val = ttest_1samp(talent_df['Satisfaction'], hyp_value1)

print(f"T-statistic: {t_stat:.3f}")
print(f"P-value: {p_val:.3f}")
```

The t-statistic is 0.830, indicating that the sample mean is only 0.83 standard errors away from the hypothesized mean of 3.
The p-value is 0.407, which is greater than the conventional threshold of 0.05. This means the difference is not statistically significant.
Therefore, we fail to reject the null hypothesis — students' satisfaction levels do not significantly differ from indifference.

#### Use the chi-square method to analyze students' perceptions of whether the course design meets the current needs of the tech industry.

where 1=strongly disagree, 2=disagree, 3=indifferent, 4=agree, and 5=strongly agree

Choosing the null hypothesis to be:

Null hypothesis: Students are equally likely to agree or disagree.

Alternative hypothesis: Students are not equally likely to agree or disagree.


We will have two groups of responses:

Agree: 3=indifferent, 4=agree, and 5=strongly agree

Disagree: 1=strongly disagree, 2=disagre

```python
talent_df = talent_df.rename(columns={'I think the course is designed in a way that meets the current tech industry needs': 'Course_perception'})
filtered_df = talent_df.dropna(subset=['Course_perception']) # dropping the missing values so as not to skew the results

disagree = [1,2]
agree = [3,4,5]

filtered_df['Perception'] = np.where(filtered_df['Course_perception'].isin(agree), 'Agree', 'Disagree')
# filtered_df.head()

# To perform the chi-squared test
observed_values = filtered_df['Perception'].value_counts()
observed_values # Agree = 115%, Disagree = 69

total = observed_values.sum()
total

# Going by the null hypothesis, then 50% should be disagreeing and 50% agree 
expected_values = [0.5 * total, 0.5 * total]

observation = [observed_values['Disagree'], observed_values['Agree']]

chi2_statistic, p_value = chisquare(f_obs=observation, f_exp=expected_values)
print(f"Chi-square: {chi2_statistic:.3f}")
print(f"P-value: {p_value:.4f}")
```

Chi-square: 11.500

P-value: 0.0007

Since the p-value < 0.05, the result is statistically significant, therefore, we reject the null hypothesis that student perceptions of the course design are equally split (50/50) between agreement and disagreement.

This suggests that students are more likely to either agree or disagree that the course design meets the current needs of the tech industry.
