#### Identify the most common programming languages that students were exposed to during their coursework.

```python
talent_df.columns
```

```python
talent_df = talent_df.rename(columns={'What programming languages were you exposed to in the school curriculum?':'Programming languages'})

# Splitting the strings in the programming languages into strings.

talent_df['Programming languages split'] = talent_df['Programming languages'].str.split(', ')
split_col = talent_df['Programming languages split'].dropna()

# Flatten the list of lists into a single flat list
import itertools
languages = list(itertools.chain.from_iterable(split_col))

common_prog_languages = pd.Series(languages).value_counts()

# Getting the top 5 programming languages

common_prog_languages.head(5)
```
